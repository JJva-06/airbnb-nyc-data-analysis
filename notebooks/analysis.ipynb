## Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import warnings
warnings.filterwarnings('ignore')

# ── Estilo SWD ——
plt.rcParams.update({
    'figure.facecolor':  'white',
    'axes.facecolor':    'white',
    'axes.spines.top':   False,
    'axes.spines.right': False,
    'axes.spines.left':  False,
    'axes.grid':         True,
    'grid.color':        '#E8E8E8',
    'grid.linewidth':    0.7,
    'font.family':       'sans-serif',
    'axes.titlepad':     12,
})

# Paleta: un color de acento + gris para contexto
ACENTO = '#7C3AED'   # violeta — el dato que quieres destacar
TEAL   = '#0F766E'   # teal — segunda serie si la necesitas
GRIS   = '#9B9B9B'   # gris — contexto, comparación
ALERTA = '#C0392B'   # rojo — solo para alertas reales

print('✓ Setup listo')
print('Paleta: ACENTO (destacar) | TEAL (segunda serie) | GRIS (contexto) | ALERTA (problemas)')
```

```python
# Dataset: Airbnb NYC 2024 (CSV local)
DATA_PATH = '../data/airbnb_nyc_2024.csv'
print('Cargando dataset Airbnb NYC...')
df = pd.read_csv(DATA_PATH)

# Limpieza básica (siempre documentada)
df_clean = df[(df['price'] > 10) & (df['price'] < 1500)].copy()

print(f'✓ Dataset cargado: {len(df):,} listings originales')
print(f'✓ Tras filtro de precio (>$10 y <$1,500): {len(df_clean):,} listings')
print(f'✓ Barrios: {sorted(df["neighbourhood_group"].dropna().unique())}')
print(f'✓ Tipos de alojamiento: {df["room_type"].unique()}')
```

## 1. Distribución de precios

**Pregunta de negocio:**  
¿Cómo se distribuyen los precios por noche en el mercado de Airbnb en NYC?

**Visualización:**  
Histograma con línea de mediana.

**¿Por qué esta visualización?**  
El histograma permite observar la distribución, asimetría y presencia de valores extremos. La mediana se utiliza como medida de referencia debido a la posible asimetría de los precios.

```python
fig, ax = plt.subplots(figsize=(11, 4.5))

ax.hist(df_clean['price'], bins=80, color=ACENTO, alpha=0.85,
        edgecolor='white', linewidth=0.3)

med = df_clean['price'].median()
mea = df_clean['price'].mean()
p75 = df_clean['price'].quantile(0.75)

ax.axvline(med, color=TEAL, lw=2.2, linestyle='--')
ax.axvline(mea, color='#B45309', lw=1.8, linestyle=':')

# Anotaciones directas — sin leyenda
ymax = ax.get_ylim()[1]
ax.text(med+5, ymax*0.88, f'Mediana\n${med:.0f}/noche', color=TEAL,
        fontsize=9.5, fontweight='bold')
ax.text(mea+5, ymax*0.72, f'Media\n${mea:.0f}/noche', color='#B45309',
        fontsize=9.5, fontweight='bold')
ax.text(df_clean['price'].quantile(0.95), ymax*0.55,
        f'La media es ${mea-med:.0f} más alta que la mediana\npor la cola de alojamientos de lujo',
        color=GRIS, fontsize=9, style='italic')

ax.set_title(
    f'El 75% de los alojamientos de Airbnb NYC cuesta ${p75:.0f}/noche o menos —\n'
    'pero la cola de alojamientos premium eleva la media artificialmente',
    fontsize=11, fontweight='bold'
)
ax.set_xlabel('Precio por noche (USD)', color=GRIS, fontsize=9)
ax.set_ylabel('N° de listings', color=GRIS, fontsize=9)
ax.tick_params(colors=GRIS)

plt.tight_layout()
plt.savefig('../outputs/figures/viz1_distribucion_precios.png', dpi=150, bbox_inches='tight')
plt.show()

# Insight para el reporte
print('═'*60)
print('INSIGHT PARA EL REPORTE:')
print(f'  Mediana: ${med:.0f}/noche  |  Media: ${mea:.0f}/noche')
print(f'  Diferencia: ${mea-med:.0f} (la media sobreestima el precio típico)')
print(f'  El 75% de listings cuesta ≤ ${p75:.0f}/noche')
print('═'*60)
print('✓ Guardado como viz1_distribucion_precios.png')
```

## 2. Ranking de barrios por precio

**Pregunta de negocio:**  
¿En qué barrios se encuentran los alojamientos más caros y cuál es la diferencia de precios entre ellos?

**Visualización:**  
Barras horizontales ordenadas de mayor a menor.

**¿Por qué esta visualización?**  
Las barras permiten comparar fácilmente magnitudes entre categorías. Ordenarlas de mayor a menor facilita identificar rápidamente los barrios con precios más altos, mientras que la orientación horizontal permite mostrar cómodamente los nombres de los barrios.

```python
precios = (df_clean.groupby('neighbourhood_group')['price']
                   .median()
                   .sort_values(ascending=True))  # ascending=True: barh dibuja de abajo hacia arriba → el mayor queda arriba

fig, ax = plt.subplots(figsize=(9, 4))

colors = [ACENTO if v == precios.max() else GRIS for v in precios.values]
bars = ax.barh(precios.index, precios.values, color=colors, edgecolor='white', linewidth=0.5, height=0.6)

for bar, val in zip(bars, precios.values):
    ax.text(val + 1.5, bar.get_y() + bar.get_height()/2,
            f'${val:.0f}', va='center', fontsize=10.5,
            color=ACENTO if val == precios.max() else GRIS,
            fontweight='bold' if val == precios.max() else 'normal')

ax.set_xlim(0, precios.max() * 1.2)
ax.grid(axis='y', visible=False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_color('#E8E8E8')
ax.tick_params(left=False, colors=GRIS)
ax.set_xlabel('Precio mediano por noche (USD)', color=GRIS, fontsize=9)

ratio = precios.max() / precios.min()
caro, barato = precios.idxmax(), precios.idxmin()
ax.set_title(
    f'{caro} es {ratio:.1f}x más caro que {barato} en precio mediano por noche\n'
    f'(mediana, no media — más representativa en distribuciones asimétricas)',
    fontsize=11, fontweight='bold'
)

plt.tight_layout()
plt.savefig('../outputs/figures/viz2_ranking_barrios.png', dpi=150, bbox_inches='tight')
plt.show()

print('═'*60)
print('INSIGHT PARA EL REPORTE:')
print(f'  {caro}: ${precios.max():.0f}  |  {barato}: ${precios.min():.0f}')
print(f'  Ratio: Manhattan es {ratio:.1f}x el precio del barrio más económico')
print('═'*60)
```

## 3. Relación entre reviews y precio

**Pregunta de negocio:**  
¿Los alojamientos más baratos acumulan más reviews? ¿Existe una relación entre el precio y la cantidad de reviews?

**Visualización:**  
Scatter plot con transparencia y línea de tendencia.

**¿Por qué esta visualización?**  
El scatter plot permite observar la relación entre dos variables cuantitativas y detectar patrones o tendencias. La transparencia facilita visualizar la concentración de puntos debido al gran número de listings analizados.

```python
base = df_clean[df_clean['number_of_reviews'] < 400]
muestra = base.sample(n=min(6000, len(base)), random_state=42)

barrio_colors = {
    'Manhattan':    ACENTO,
    'Brooklyn':     TEAL,
    'Queens':       '#B45309',
    'Bronx':        ALERTA,
    'Staten Island': GRIS,
}

fig, ax = plt.subplots(figsize=(11, 5))

for barrio, grp in muestra.groupby('neighbourhood_group'):
    ax.scatter(grp['number_of_reviews'], grp['price'],
               color=barrio_colors.get(barrio, GRIS),
               alpha=0.15, s=14, label=barrio)

# Línea de tendencia
m, c = np.polyfit(base['number_of_reviews'], base['price'], 1)  # [pendiente, intercepto]
x_line = np.linspace(0, base['number_of_reviews'].max(), 100)
ax.plot(x_line, c + m*x_line, color='#111111', lw=2, linestyle='--', alpha=0.6, label='Tendencia')

# Anotación del insight
share_mod = (base.loc[base['price'].between(50, 150, inclusive='right'), 'number_of_reviews'].sum()
             / base['number_of_reviews'].sum() * 100)
corr = base[['price','number_of_reviews']].corr().iloc[0,1]
ax.text(0.70, 0.95, f'Correlación: r = {corr:.2f}\n(negativa débil — a mayor precio,\nligeramente menos reviews)',
        fontsize=9, color='#111111', transform=ax.transAxes, va='top',
        bbox=dict(boxstyle='round,pad=0.4', facecolor='white', edgecolor='#E8E8E8', alpha=0.9))

ax.set_title(
    f'Los alojamientos de precio moderado ($50-150) acumulan el {share_mod:.0f}% de las reviews —\n'
    'los más caros tienden a tener menos reseñas',
    fontsize=11, fontweight='bold'
)
ax.set_xlabel('Número de reviews', color=GRIS, fontsize=9)
ax.set_ylabel('Precio por noche (USD)', color=GRIS, fontsize=9)
ax.set_ylim(bottom=0)
ax.legend(title='Barrio', fontsize=8, title_fontsize=8, framealpha=0.8)

plt.tight_layout()
plt.savefig('../outputs/figures/viz3_reviews_vs_precio.png', dpi=150, bbox_inches='tight')
plt.show()

print('═'*60)
print('INSIGHT PARA EL REPORTE:')
print(f'  Correlación precio vs. reviews: r = {corr:.3f}')
print(f'  Listings de $50-150 acumulan el {share_mod:.1f}% de las reviews')
print('  Relación negativa débil: los alojamientos más caros acumulan menos reviews.')
print('  Interpretación posible: mayor rotación de huéspedes en el segmento económico.')
print('═'*60)
```

## 4. Composición por tipo de alojamiento

**Pregunta de negocio:**  
¿La composición de los tipos de alojamiento varía según el barrio?

**Visualización:**  
Barras apiladas al 100%.

**¿Por qué esta visualización?**  
Las barras normalizadas permiten comparar la proporción de cada tipo de alojamiento entre barrios, independientemente de la cantidad total de listings. Esto evita que los barrios con mayor número de alojamientos dominen la comparación.

```python
room_map = {
    'Entire home/apt': 'Casa/Depto completo',
    'Private room':    'Habitación privada',
    'Shared room':     'Hab. compartida',
    'Hotel room':      'Hotel',
}
df_clean['room_tipo'] = df_clean['room_type'].map(room_map).fillna('Otro')

pivot = df_clean.groupby(['neighbourhood_group','room_tipo']).size().unstack(fill_value=0)
pivot_pct = pivot.div(pivot.sum(axis=1), axis=0) * 100

# Ordenar por % de casas completas
orden = pivot_pct['Casa/Depto completo'].sort_values(ascending=True).index
pivot_pct = pivot_pct.loc[orden]

orden_tipos = [t for t in ['Casa/Depto completo', 'Habitación privada', 'Hotel', 'Hab. compartida', 'Otro']
               if t in pivot_pct.columns]
pivot_pct = pivot_pct[orden_tipos]

colores_tipo = {
    'Casa/Depto completo': ACENTO,
    'Habitación privada':  TEAL,
    'Hotel':               '#B45309',
    'Hab. compartida':     GRIS,
    'Otro':                '#E8E8E8',
}

lider_casa = pivot_pct['Casa/Depto completo'].idxmax()
lider_hab  = pivot_pct['Habitación privada'].idxmax()

fig, ax = plt.subplots(figsize=(11, 4.5))
bottom = np.zeros(len(pivot_pct))

for col in pivot_pct.columns:
    vals = pivot_pct[col].values
    bars = ax.bar(pivot_pct.index, vals, bottom=bottom,
                  color=colores_tipo.get(col, GRIS),
                  label=col, edgecolor='white', linewidth=0.8)
    for bar, v, b in zip(bars, vals, bottom):
        if v > 9:
            ax.text(bar.get_x()+bar.get_width()/2, b+v/2,
                    f'{v:.0f}%', ha='center', va='center',
                    fontsize=9, color='white', fontweight='bold')
    bottom += vals

ax.set_ylim(0, 100)
ax.set_ylabel('% de listings', color=GRIS, fontsize=9)
ax.tick_params(colors=GRIS)
ax.legend(loc='upper left', bbox_to_anchor=(1.01, 1), fontsize=8, frameon=False)
ax.grid(axis='x', visible=False)

ax.set_title(
    f'{lider_casa} lidera en casas/deptos completos ({pivot_pct.loc[lider_casa, "Casa/Depto completo"]:.0f}%) — '
    f'{lider_hab} tiene la mayor\nproporción de habitaciones privadas ({pivot_pct.loc[lider_hab, "Habitación privada"]:.0f}%)',
    fontsize=11, fontweight='bold'
)

plt.tight_layout()
plt.savefig('../outputs/figures/viz4_composicion_tipo.png', dpi=150, bbox_inches='tight')
plt.show()

print('═'*60)
print('INSIGHT PARA EL REPORTE:')
print(pivot_pct[['Casa/Depto completo','Habitación privada']].round(1).to_string())
print('═'*60)
```

## 5. Concentración del mercado

**Pregunta de negocio:**  
¿El mercado de Airbnb en NYC está concentrado en pocos hosts o se encuentra distribuido entre muchos?

**Visualización:**  
Curva de concentración tipo Lorenz.

**¿Por qué esta visualización?**  
La curva de Lorenz permite analizar cómo se distribuyen los listings entre los hosts y determinar si una pequeña proporción de ellos concentra una gran parte del mercado. La distancia respecto a la línea de igualdad perfecta permite visualizar el nivel de concentración.

```python
host_counts = df_clean.groupby('host_id').size().sort_values()
total_hosts    = len(host_counts)
total_listings = host_counts.sum()

cum_hosts    = np.arange(1, total_hosts+1) / total_hosts * 100
cum_listings = host_counts.cumsum() / total_listings * 100

def pct_listings_hasta(pct_hosts):
    """% de listings en manos del pct_hosts% de hosts más pequeños."""
    return cum_listings.values[int(total_hosts * pct_hosts / 100) - 1]

fig, ax = plt.subplots(figsize=(9, 5.5))

ax.plot(cum_hosts, cum_listings.values, color=ACENTO, lw=2.5, label='Distribución real')
ax.plot([0,100], [0,100], color=GRIS, lw=1.2, linestyle='--', alpha=0.6, label='Igualdad perfecta')
ax.fill_between(cum_hosts, cum_listings.values, cum_hosts, alpha=0.08, color=ACENTO)

# Anotaciones de insights clave
for pct_host, label_text, color in [(80, '80% de hosts', GRIS), (95, '95% de hosts', '#B45309'), (99, '99% de hosts', ALERTA)]:
    pct_list = pct_listings_hasta(pct_host)
    ax.scatter(pct_host, pct_list, color=color, s=60, zorder=5)
    offset_x = -18 if pct_host < 90 else -22
    ax.annotate(
        f'{label_text}\n= {pct_list:.0f}% listings',
        xy=(pct_host, pct_list), xytext=(pct_host+offset_x, pct_list-12),
        fontsize=8.5, color=color, fontweight='bold',
        arrowprops=dict(arrowstyle='->', color=color, lw=1)
    )

pct_top1 = 100 - pct_listings_hasta(99)
pct_hosts_chicos = (host_counts <= 2).mean() * 100

ax.set_title(
    f'El mercado de Airbnb NYC está altamente concentrado:\n'
    f'el 1% de hosts más grandes controla el {pct_top1:.0f}% de los listings',
    fontsize=11, fontweight='bold'
)
ax.set_xlabel('% acumulado de hosts (de menor a mayor tamaño)', color=GRIS, fontsize=9)
ax.set_ylabel('% acumulado de listings', color=GRIS, fontsize=9)
ax.legend(fontsize=9, framealpha=0.8)
ax.set_xlim(0, 101); ax.set_ylim(0, 101)

plt.tight_layout()
plt.savefig('../outputs/figures/viz5_concentracion_mercado.png', dpi=150, bbox_inches='tight')
plt.show()

print('═'*60)
print('INSIGHT PARA EL REPORTE:')
print(f'  El 1% de hosts más grandes controla el {pct_top1:.0f}% del total de listings.')
print(f'  El {pct_hosts_chicos:.0f}% de los hosts tiene 1-2 listings.')
print(f'  El 80% de hosts más pequeños controla solo el {pct_listings_hasta(80):.0f}% del mercado.')
print('  → Mercado altamente concentrado con pocos operadores profesionales grandes.')
print('═'*60)
```
